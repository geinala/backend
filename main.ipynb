```python
def create_data_model(time_matrix_data):
    """
    Create data model for OR-Tools using time matrix

    Assumptions:
    - time_matrix_data from OSRM is in SECONDS (float)
    - OR-Tools requires INTEGER values
    - scaling_factor is used only if higher precision is needed
      (default = 1, meaning no scaling)
    """

    data = {}

    # Konversi ke integer (OR-Tools requirement)
    data['time_matrix'] = [
        [int(round(t)) for t in row]
        for row in time_matrix_data
    ]

    return data
```

```python
def create_routing_model(data):
    """
    Create routing manager and model
    
    Parameters:
    - data: data dictionary from create_data_model
    
    Returns:
    - manager: RoutingIndexManager
    - routing: RoutingModel
    """
    manager = pywrapcp.RoutingIndexManager(
        len(data["time_matrix"]),
        data["num_vehicles"],
        data["depot"]
    )
    
    print("Manager created with {} locations, {} vehicles, depot at {}".format(
        len(data["time_matrix"]),
        data["num_vehicles"],
        data["depot"]
    ))
    
    routing = pywrapcp.RoutingModel(manager)
    
    return manager, routing
```

```python
def extract_solution(data, manager, routing, solution): 
    routes = []
    total_distance = 0
    total_load = 0
    
    for vehicle_id in range(data['num_vehicles']):
        index = routing.Start(vehicle_id)
        route_load = 0
        route_dist = 0
        route_nodes = []
        
        while not routing.IsEnd(index):
            node_index = manager.IndexToNode(index)
            route_load += data['demands'][node_index]
            route_nodes.append(node_index)
            
            previous_index = index
            index = solution.Value(routing.NextVar(index))
            route_dist += routing.GetArcCostForVehicle(previous_index, index, vehicle_id)
            
        routes.append({
            "vehicle_id": vehicle_id,
            "route": route_nodes,
            "load": route_load,
            "distance": route_dist
        })
        total_distance += route_dist
        total_load += route_load
        
    return {"routes": routes, "total_distance": total_distance, "total_load": total_load}
```

```python
def solve_vrp_snapshot(data_input):
    data_input['time_matrix'] = create_data_model(data_input['time_matrix'])['time_matrix']
    
    manager, routing = create_routing_model(data_input)
    
    def distance_callback(from_index, to_index):
        from_node = manager.IndexToNode(from_index)
        to_node = manager.IndexToNode(to_index)
        return data_input['time_matrix'][from_node][to_node]

    def demand_callback(from_index):
        from_node = manager.IndexToNode(from_index)
        return data_input['demands'][from_node]
    
    transit_callback_index = routing.RegisterTransitCallback(distance_callback)
    routing.SetArcCostEvaluatorOfAllVehicles(transit_callback_index)
    
    demand_callback_index = routing.RegisterUnaryTransitCallback(demand_callback)
    
    routing.AddDimensionWithVehicleCapacity(
        demand_callback_index, 0,
        data_input['vehicle_capacities'],
        True,
        'Capacity'
    )
    
    search_parameters = pywrapcp.DefaultRoutingSearchParameters()
    
    search_parameters.first_solution_strategy = (
        routing_enums_pb2.FirstSolutionStrategy.PARALLEL_CHEAPEST_INSERTION
    )
    
    search_parameters.local_search_metaheuristic = (
        routing_enums_pb2.LocalSearchMetaheuristic.TABU_SEARCH
    )
    
    search_parameters.time_limit.seconds = 5
    
    solution = routing.SolveWithParameters(search_parameters)
    
    if solution:
        return extract_solution(data_input, manager, routing, solution)
    else:
        return None
```
